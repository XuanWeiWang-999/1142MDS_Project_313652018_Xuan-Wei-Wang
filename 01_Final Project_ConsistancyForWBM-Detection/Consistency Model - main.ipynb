```python
# ============================================================
# WM811K Dataset Loading + Visualization
# MDS Final Project
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Load dataset
# ------------------------------------------------------------

# 修改成你的路徑
data_path = "C:\\Users\\user\\Downloads\\LSWMD.pkl"

df = pd.read_pickle(data_path)

print(df.head())
print(df.columns)

# ------------------------------------------------------------
# Basic information
# ------------------------------------------------------------

print("Dataset shape:", df.shape)

print("\nFailure type count:")
print(df['failureType'].value_counts())

# ------------------------------------------------------------
# Remove unlabeled data
# ------------------------------------------------------------

df = df[df['failureType'].notna()]

print("\nAfter removing NaN:")
print(df.shape)

# ------------------------------------------------------------
# Show wafer map examples
# ------------------------------------------------------------

def plot_wafer(index):

    wafer = df.iloc[index]['waferMap']
    label = df.iloc[index]['failureType'][0]

    plt.figure(figsize=(4,4))
    plt.imshow(wafer, cmap='gray')
    plt.title(label)
    plt.axis('off')
    plt.show()

# ------------------------------------------------------------
# Show several samples
# ------------------------------------------------------------

for i in range(5):

    plot_wafer(i)

# ------------------------------------------------------------
# Show one sample per class
# ------------------------------------------------------------

classes = df['failureType'].apply(lambda x: x[0]).unique()

print("\nClasses:")
print(classes)

for c in classes:

    idx = df[df['failureType'].apply(lambda x: x[0]) == c].index[0]

    wafer = df.loc[idx, 'waferMap']

    plt.figure(figsize=(4,4))
    plt.imshow(wafer, cmap='gray')
    plt.title(c)
    plt.axis('off')
    plt.show()
```
